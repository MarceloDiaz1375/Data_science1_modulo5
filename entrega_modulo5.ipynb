```python
# ==========================================
# 1. INICIALIZACIÓN Y CARGA DE DATOS
# ==========================================
import pandas as pd

# Carga del dataset desde el archivo Excel
df = pd.read_excel('base_hardware_2.xlsx')

# Inspección inicial del volumen y estructura
print(f"Dimensiones iniciales del dataset: {df.shape}")
display(df.head())
df.info()
```

```python
# ==========================================
# 2. DIAGNÓSTICO TÉCNICO
# ==========================================
print("--- Conteo de Valores Nulos por Columna ---")
print(df.isnull().sum())

print("\n--- Resumen Estadístico ---")
display(df.describe())
```

```python
df.describe()
```

```python
# ==========================================
# BLOQUE: LIMPIEZA RIGUROSA DE DATOS (CORREGIDO)
# ==========================================
import pandas as pd

# 1. Verificación y eliminación de registros duplicados
duplicados_antes = df.duplicated().sum()
df.drop_duplicates(inplace=True)
print(f"Registros duplicados eliminados: {duplicados_antes}")

# 2. Imputación inteligente de PRECIO UNITARIO
precios_referencia = df.groupby('nombre_producto')['precio_unitario'].median().to_dict()

def rellenar_precio_unitario(row):
    if pd.isna(row['precio_unitario']):
        return precios_referencia.get(row['nombre_producto'], row['precio_unitario'])
    return row['precio_unitario']

df['precio_unitario'] = df.apply(rellenar_precio_unitario, axis=1)


# 3. Imputación de DESCUENTO PORCENTAJE basada en la Regla de Negocio
def aplicar_regla_descuento(row):
    if pd.isna(row['descuento_porcentaje']):
        cant = row['cantidad']
        if cant == 3:
            return 0.10
        elif cant == 4:
            return 0.15
        elif cant >= 5:
            return 0.20
        else:
            return 0.00
    return row['descuento_porcentaje']

df['descuento_porcentaje'] = df.apply(aplicar_regla_descuento, axis=1)


# 4. Tratamiento del resto de nulos (Asignación directa sin FutureWarning)
df['dias_demora'] = df['dias_demora'].fillna(df['dias_demora'].median())
df['genero_cliente'] = df['genero_cliente'].fillna('Desconocido')
df['provincia_cliente'] = df['provincia_cliente'].fillna('Desconocida')


# 5. Conversión estricta de tipos de datos (Fechas)
df['fecha_compra'] = pd.to_datetime(df['fecha_compra'])
df['fecha_entrega'] = pd.to_datetime(df['fecha_entrega'])


# 6. Resetear el índice tras la limpieza
df.reset_index(drop=True, inplace=True)

print("\n--- ¡Limpieza rigurosa completada con éxito! ---")
print(f"Total de valores nulos restantes en el DataFrame: {df.isnull().sum().sum()}")
```

```python
# ==========================================
# 4. AGREGACIONES Y ANÁLISIS DE NEGOCIO
# ==========================================

# Agregación 1: Promedio de ventas (monto_total) por categoría de producto
print("1. Promedio de Monto Total por Categoría de Producto:")
gb_categoria = df.groupby('categoria_producto')['monto_total'].mean().reset_index()
display(gb_categoria)
print("-" * 50)

# Agregación 2: Conteo de ventas y facturación acumulada por provincia
print("2. Desempeño Comercial y Conteo de Operaciones por Provincia:")
gb_provincia = df.groupby('provincia_cliente').agg(
    total_operaciones=('id_venta', 'count'),
    facturacion_acumulada=('monto_total', 'sum')
).reset_index()
display(gb_provincia)
print("-" * 50)

# Agregación 3: Promedio de días de demora logística según el medio de pago
print("3. Promedio de Días de Demora según el Medio de Pago:")
gb_pago = df.groupby('medio_pago')['dias_demora'].mean().reset_index()
display(gb_pago)
```
